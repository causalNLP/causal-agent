# Healthcare Analysis Tutorial: Effect of Medical Treatment on Patient Recovery

This tutorial demonstrates how to use CAIS to analyze the causal effect of medical interventions on patient outcomes. We'll use a hospital treatment dataset to explore how a new medical treatment affects patient recovery times.

```python
# Import required libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from causal_agent import run_causal_analysis
import os

print("Healthcare analysis tutorial setup complete!")
```

## Data Loading and Exploration

Let's load the hospital treatment dataset and explore its structure.

```python
# Create simulated healthcare dataset
np.random.seed(42)
n = 1000

df = pd.DataFrame({
    'patient_id': range(n),
    'age': np.random.normal(65, 15, n),
    'severity_score': np.random.exponential(2, n),
    'new_treatment': np.random.binomial(1, 0.4, n),
    'recovery_days': np.random.normal(15, 5, n)
})

print(f"Dataset created with {len(df)} patients")
print(df.head())
```

## Running CAIS Analysis

Now let's use CAIS to analyze the causal effect of the medical treatment.

```python
# Simulate CAIS analysis results
print("Running CAIS analysis...")

# Mock results for demonstration
result = {
    'results': {
        'results': {
            'method_used': 'Propensity Score Matching',
            'effect_estimate': -2.5,
            'standard_error': 0.4,
            'p_value': 0.001
        }
    }
}

print(f"Method: {result['results']['results']['method_used']}")
print(f"Effect: {result['results']['results']['effect_estimate']} days")
print("Analysis complete!")
```

```python
# Visualize the treatment effect
plt.figure(figsize=(10, 6))
treatment_groups = df.groupby('new_treatment')['recovery_days'].mean()
plt.bar(['Control', 'Treatment'], treatment_groups.values, color=['lightcoral', 'lightblue'])
plt.title('Average Recovery Days by Treatment Group')
plt.ylabel('Recovery Days')
plt.show()

print(f"Treatment effect: {treatment_groups[1] - treatment_groups[0]:.2f} days")
```

```python
# Summary statistics by treatment group
print("Summary Statistics by Treatment Group:")
print("\nControl Group:")
print(df[df['new_treatment']==0][['age', 'severity_score', 'recovery_days']].describe())
print("\nTreatment Group:")
print(df[df['new_treatment']==1][['age', 'severity_score', 'recovery_days']].describe())
```
