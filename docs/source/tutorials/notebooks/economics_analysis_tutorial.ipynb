This tutorial demonstrates how to use CAIS to analyze economic policy effects using difference-in-differences methodology. We'll examine how minimum wage increases affect employment levels.

```python
# Import required libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from causal_agent import run_causal_analysis
import os

print("Economics analysis tutorial setup complete!")
```

## Data Loading and Exploration

Let's create a simulated minimum wage dataset for difference-in-differences analysis.

```python
# Create simulated economics dataset
np.random.seed(42)

# Panel data: 20 states over 10 years
states = [f"State_{i:02d}" for i in range(1, 21)]
years = list(range(2010, 2020))

data = []
for state in states:
    for year in years:
        treated = 1 if state in states[:10] and year >= 2015 else 0
        employment = 65 + np.random.normal(0, 2) - 1.5 * treated
        
        data.append({
            'state': state,
            'year': year,
            'employment_rate': employment,
            'treatment': treated
        })

df = pd.DataFrame(data)
print(f"Dataset created with {len(df)} observations")
print(df.head())
```

## Visualizing the Natural Experiment

Let's visualize the difference-in-differences setup.

```python
# Create DiD visualization
treated_states = states[:10]
treated_trend = df[df['state'].isin(treated_states)].groupby('year')['employment_rate'].mean()
control_trend = df[~df['state'].isin(treated_states)].groupby('year')['employment_rate'].mean()

plt.figure(figsize=(10, 6))
plt.plot(treated_trend.index, treated_trend.values, 'b-o', label='Treated States', linewidth=2)
plt.plot(control_trend.index, control_trend.values, 'r-o', label='Control States', linewidth=2)
plt.axvline(x=2015, color='gray', linestyle='--', alpha=0.7, label='Policy Change')
plt.title('Employment Rate Trends')
plt.xlabel('Year')
plt.ylabel('Employment Rate (%)')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

## Running CAIS Analysis

Now let's use CAIS to analyze the causal effect using difference-in-differences.

```python
# Simulate CAIS analysis results
print("Running CAIS analysis...")

# Mock results for demonstration
result = {
    'results': {
        'results': {
            'method_used': 'Difference-in-Differences',
            'effect_estimate': -1.5,
            'standard_error': 0.3,
            'p_value': 0.001
        }
    }
}

print(f"Method: {result['results']['results']['method_used']}")
print(f"Effect: {result['results']['results']['effect_estimate']} percentage points")
print("Analysis complete!")
```

## Economic Interpretation

The results show that minimum wage increases reduced employment by 1.5 percentage points, which is consistent with standard economic theory.

```python
# Calculate manual DiD estimate
pre_treatment = df[df['year'] < 2015]
post_treatment = df[df['year'] >= 2015]

treated_states = states[:10]
treated_pre = pre_treatment[pre_treatment['state'].isin(treated_states)]['employment_rate'].mean()
treated_post = post_treatment[post_treatment['state'].isin(treated_states)]['employment_rate'].mean()
control_pre = pre_treatment[~pre_treatment['state'].isin(treated_states)]['employment_rate'].mean()
control_post = post_treatment[~post_treatment['state'].isin(treated_states)]['employment_rate'].mean()

did_effect = (treated_post - treated_pre) - (control_post - control_pre)
print(f"Manual DiD estimate: {did_effect:.2f} percentage points")
```
